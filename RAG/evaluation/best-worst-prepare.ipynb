## Best worst scaling 

```python
from datasets import load_from_disk
import pandas as pd
```

```python
answers = load_from_disk("all_answers")
```

```python
answers = answers.to_pandas()
```

```python
subset = answers[(answers['question_mapped']=="Question 1")]
```

```python
all_answers = []
for i, row in subset.iterrows():
    q = row["question_mapped"].split(" ")[1]
    lang = row["language"]
    model = row["model"].split("-")[1]
    for j, a in enumerate(row["answers"]):
        #questionnum_repeat_lang_model_setting
        all_answers.append(f"#{q}_{j}_{lang}_{model}_rag#{a}")
    for j, a in enumerate(row["baseline_answer"]):
        #questionnum_repeat_lang_model_setting
        all_answers.append(f"#{q}_{j}_{lang}_{model}_base#{a}")        
```

```python
for lang in ["EN", "FR", "NL"]:
    subset = [x for x in all_answers if lang in x.split("#")[1]]
    with open(f'best-worst/best-worst-{lang}.txt', 'w', encoding="utf-8") as f:
        for line in subset:
            f.write(f"{line}\n")
```
